# Finding Duplicate Records

```python
duplicate = df[df.duplicated()]

print(duplicate.head(10))
```

```python
print("total no. of duplicated rows are:",df.duplicated().sum())
```

```python
print(df[df.duplicated(keep=False)].head(20))
```

```python
df_no_duplicates = df.drop_duplicates()

print("Original dataset:", len(df))
print("After removing duplicates:", len(df_no_duplicates)) # total no. of of non-duplicated rows are thre in the dataframe..
print("Duplicates removed:", len(df) - len(df_no_duplicates))
```

### Observation

The dataset originally contained 541,909 records.
After identifying and removing exact duplicate rows, 536,641
records remained.

A total of 5,268 duplicate records were identified.

These duplicate rows may unnecessarily affect sales calculations
and other statistical analysis, so they will be removed from the
working dataset after validation.

```python
print("duplicated before:",df.duplicated().sum())
print("duplicated after:",df_no_duplicates.duplicated().sum())
```

## Negative quantity analysis

```python
cancelled_orders = df[
    df["InvoiceNo"].str.startswith("C")
]

print("Total cancelled invoice records:", len(cancelled_orders))
print(cancelled_orders.head(10))
```

```python
# just checking in invoice no column how many "C" letter is there which means cancelled order.
order = df[df["InvoiceNo"].str.startswith("C")]

print(order.head(10))
```

```python
# Compare negative quantities with cancellation invoices

print(
    "Negative quantity records:",
    (df["Quantity"] < 0).sum()
)

print(
    "Cancellation invoice records:",
    df["InvoiceNo"].str.startswith("C").sum()
)

print(
    "Negative quantities with cancellation invoices:",
    ((df["Quantity"] < 0) &
     (df["InvoiceNo"].str.startswith("C"))).sum()
)
```
